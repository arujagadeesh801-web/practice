```python
import numpy as np # numerical operation
import pandas as pd # data manipulation
import matplotlib.pyplot as plt # data visualization
```

#### Data Loading

```python
data = pd.read_csv("order.csv")
data.head()
```

#### Questio1 : which payment method is higly prefered by customers

```python
data["payment_method"].value_counts()
```

```python
data["payment_method"].value_counts().plot(
                                      kind = "pie",
                                      autopct="%.2f%%",
                                      explode =(0.1,0,0,0)
                                      )
```

#### Question2: Payment_method by city wise

```python
(data.groupby("city_user")["payment_method"].value_counts()
                                            .reset_index()
)
```

#### Question 3: Total revenue by city

```python
(data.groupby("city_user")["total_amount"].sum()
    .sort_values(ascending = False)
)
```

```python
(data.groupby("city_user")["total_amount"].sum()
    .sort_values(ascending = False)
    .plot(kind = "barh",
          color="skyblue"
         )
)
plt.title("City wise revenue")
plt.xlabel("Renvue")
plt.ylabel("City")
plt.show()
```

<h3 style="color:red"><i>File handling</i></h3>

```python
'''
file operation modes:
'w'- write
'r' - read
'a' - append
'w+' - write and read
'r+' - read and write
'x'  - open file in exclusive mode( throw error if file alread exists)
'rb' - reading the byte data
'wb' - writing byte data

File operation method:
1. write()
2. writelines()
3. read()
4. readlines()
5. seek() - pointer movement
6. tell() - pointer position
'''
```

```python
with open("sample.txt","a") as file:
    file.write("\nthis is sample text")
```

```python
with open("sample.txt","w") as file:
    file.writelines('''
    okay everthing is cool here
    file handling is good to use 
    when we have less data
    ''')
```

```python
with open("sample.txt","r") as file:
    data = file.read()
    print(data)
```

```python
with open("demo1.txt","w+") as file:
    file.writelines('''
    Hy there,
    this is sample text file!
    Hurray!!!!!
    ''')
    file.seek(0) # point to current cursor!!!!
    data = file.read()
print(data)
```

```python
file = open("sample.txt",'r')
data = file.read()
file.close()
print(data)
```

```python
with open("sample.txt","r") as file:
    file.seek(20) # set pointer to character 20
    print(file.tell()) # get the pointer location 
```

#### Exception Handling

```python
'''
Exception  - logical error 
Error - Syntax error

ZeroDevisionException
IndexRangeoutofBound

'''
```

```python
try: # used to capture the error
    l = [1,2,3,4]
    print(l[0])
    a = 10
    b = 0
    print(a/b)
except IndexError:
    print("Index out of range")
except Exception as e: # used to handle the error
    print(e)
else: # execute only exception not occured!
    print("Else block excuted")
finally: # no matter what this code will excute
    print("Finally code!!!")
print("Program done")
```
