```python
import pandas as pd
import os

filename = "students.csv"

# CSV file create
if not os.path.exists(filename):
    df = pd.DataFrame(columns=[
        "Student_ID","Name","Tamil","English","Maths",
        "Science","Social","Total","Average","Grade"
    ])
    df.to_csv(filename, index=False)

# Student input
student_id = input("Enter Student ID: ")
name = input("Enter Student Name: ")

tamil = int(input("Enter Tamil Mark: "))
english = int(input("Enter English Mark: "))
maths = int(input("Enter Maths Mark: "))
science = int(input("Enter Science Mark: "))
social = int(input("Enter Social Mark: "))

# Calculation
total = tamil + english + maths + science + social
average = total / 5

# Grade calculation
if average >= 90:
    grade = "A+"
elif average >= 80:
    grade = "A"
elif average >= 70:
    grade = "B"
elif average >= 60:
    grade = "C"
elif average >= 50:
    grade = "D"
else:
    grade = "F"

# Save data
student_data = pd.DataFrame([{
    "Student_ID": student_id,
    "Name": name,
    "Tamil": tamil,
    "English": english,
    "Maths": maths,
    "Science": science,
    "Social": social,
    "Total": total,
    "Average": average,
    "Grade": grade
}])

student_data.to_csv(filename, mode="a", header=False, index=False)

print("\nStudent Data Saved Successfully!")
print("--------------------------------")
print(student_data)

# Display all records
print("\nAll Student Records:")
print(pd.read_csv(filename))
```
